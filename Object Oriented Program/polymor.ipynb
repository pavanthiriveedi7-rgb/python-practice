Polymorphism:

What is polymorphism:The word polymorphism means having many forms. in programming,polymorphism means the same function name (but different signatures) being used for different types.the key difference is the data types and numbers of arguments used i  function.

METHOD OVERRIDING:
an object-oriented programming feature where a subclass provides its own specific implementation for a method that is already defined in its superclass, keeping the same name, parameters, and return type. It allows child classes to customize inherited behavior, enabling runtime polymorphism (dynamic method dispatch), where the appropriate method version is chosen at runtime based on the actual object's type, not just the reference.

```python
class marks:
  def total(self):
    print("Students total percentage is 70 % with '1' backlog")
class final_marks(marks):
  def total_marks(self,backlogs):
    self.backlogs = backlogs
    if backlogs == 0:
      print("Students total percentage is 80 % with '0' backlog")
    else:
      super().total()
```

```python
pavan = marks()
```

```python
pavan_ = final_marks()
```

```python
pavan_.total_marks(8)
```

```python
class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Polymorphic behavior
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
```

```python
class student:
    def clg_fees(self):
       print("students per year 50000 fees")
class sem_fees(student):
    def clg_fees(self,fine):
      self.fine = fine
      super().clg_fees()
      total_fees = 50000 + fine
      print("to pay for this sem fees is ",total_fees)


```

```python
regular = student()
```

```python
irregular = sem_fees()
```

```python
regular.clg_fees()
```

```python
irregular.clg_fees(1000)
```

```python
irregular.clg_fees(500)
```

Methos overloading:

Python does not support method overloading in the traditional sense, unlike languages such as Java or C++. In Python, if you define multiple methods with the same name within a single class, the last definition will override the previous ones.

```python
#Examples

name = "venkata"
name = "pavan"
name = "kumar"

name = "cho"
print(name)
```
